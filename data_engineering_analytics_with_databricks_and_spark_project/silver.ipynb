```python
%sql
-- Create Silver tables
USE CATALOG store_sales_catalog;
USE SCHEMA silver;

CREATE TABLE IF NOT EXISTS stores_silver (
  store_key BIGINT GENERATED ALWAYS AS IDENTITY,  -- Surrogate key
  store_nbr INT,                                   -- Natural key
  city STRING,
  state STRING,
  type STRING,
  cluster INT,
  _bronze_source STRING,
  created_at TIMESTAMP,
  updated_at TIMESTAMP,
  PRIMARY KEY (store_key)
) USING DELTA;

CREATE TABLE IF NOT EXISTS family_silver (
  family_key BIGINT GENERATED ALWAYS AS IDENTITY,  -- Surrogate key
  family_name STRING,                               -- Natural key (from train.family)
  created_at TIMESTAMP,
  updated_at TIMESTAMP,
  PRIMARY KEY (family_key)
) USING DELTA;

CREATE TABLE IF NOT EXISTS date_silver (
  date_key INT,                    -- YYYYMMDD format (e.g., 20170101)
  date DATE,                       -- Actual date
  day_of_week INT,
  week_of_year INT,
  month INT,
  month_name STRING,
  year INT,
  is_weekend BOOLEAN,
  day_of_month INT,
  day_name STRING,
  created_at TIMESTAMP,
  PRIMARY KEY (date_key)
) USING DELTA;

CREATE TABLE IF NOT EXISTS oil_silver (
  oil_key BIGINT GENERATED ALWAYS AS IDENTITY,  -- Surrogate key
  date DATE,                                    -- Natural key Converted from STRING to DATE
  dcoilwtico DOUBLE,                            -- Drop Null Rows
  created_at TIMESTAMP,
  updated_at TIMESTAMP,
  PRIMARY KEY (date)
) USING DELTA;

CREATE TABLE IF NOT EXISTS transactions_silver (
  date DATE,                       -- Converted from STRING to DATE
  store_nbr INT,
  transactions INT,
  created_at TIMESTAMP,
  updated_at TIMESTAMP,
  PRIMARY KEY (date, store_nbr)
) USING DELTA;

CREATE TABLE IF NOT EXISTS holidays_events_silver (
  holiday_key BIGINT GENERATED ALWAYS AS IDENTITY,
  date DATE,                       -- Converted from STRING to DATE
  type STRING,                     -- Cleaned/standardized
  locale STRING,                   -- Cleaned/standardized
  locale_name STRING,              -- Cleaned/standardized
  description STRING,
  transferred BOOLEAN,             -- Converted from STRING to BOOLEAN
  is_national BOOLEAN,             -- Derived: locale = 'National'
  is_regional BOOLEAN,             -- Derived: locale = 'Regional'
  is_local BOOLEAN,                -- Derived: locale = 'Local'
  created_at TIMESTAMP,
  updated_at TIMESTAMP,
  PRIMARY KEY (holiday_key)
) USING DELTA;

CREATE TABLE IF NOT EXISTS sales_silver (
  sales_id BIGINT GENERATED ALWAYS AS IDENTITY,
  date DATE,                       -- Converted from STRING to DATE
  sales DOUBLE,                    -- Cleaned (handle negatives/nulls)
  family STRING,
  store_nbr INT,
  onpromotion INT,                 -- Cleaned nulls
  is_promotion BOOLEAN,            -- Derived: onpromotion > 0
  created_at TIMESTAMP,
  updated_at TIMESTAMP,
  PRIMARY KEY (sales_id)
) USING DELTA;

```

Silver Layer Transformations

1. Data Type Conversions: Convert all date STRING columns to DATE type
2. Null Handling: Fill or flag null values in oil prices, sales, transactions
3. Boolean Standardization: Convert "True"/"False" strings to actual BOOLEAN
4. Surrogate Keys: Create auto-increment keys for dimensions (store_key, family_key, etc.)
5. Date Dimension: Generate a complete date dimension covering all dates in your dataset
6. Deduplication: Remove any duplicate rows from bronze
7. Business Rules: Apply any domain-specific rules (e.g., negative sales = 0)

```python
%sql
-- Incremental merge from bronze to silver for stores table
-- Performs NULL checks
-- Performs Deduped merge
MERGE INTO store_sales_catalog.silver.stores_silver AS tgt
USING (
  SELECT
    store_nbr,
    city,
    state,
    type,
    cluster,
    _source_file,
    current_timestamp() AS created_at,
    current_timestamp() AS updated_at
  FROM (
    SELECT *,
      ROW_NUMBER() OVER (
        PARTITION BY store_nbr
        ORDER BY current_timestamp()
      ) AS rn
    FROM store_sales_catalog.bronze.stores_bronze
    WHERE store_nbr IS NOT NULL
  ) sub
  WHERE rn = 1
) AS src
ON tgt.store_nbr = src.store_nbr
WHEN MATCHED THEN
  UPDATE SET
    tgt.city = src.city,
    tgt.state = src.state,
    tgt.type = src.type,
    tgt.cluster = src.cluster,
    tgt._bronze_source = src._source_file,
    tgt.updated_at = current_timestamp()
WHEN NOT MATCHED THEN
  INSERT (
    store_nbr,
    city,
    state,
    type,
    cluster,
    _bronze_source,
    created_at,
    updated_at
  )
  VALUES (
    src.store_nbr,
    src.city,
    src.state,
    src.type,
    src.cluster,
    src._source_file,
    src.created_at,
    src.updated_at
  )
```

```python
%sql
-- Incremental merge from bronze to silver for family table
-- Performs NULL checks
-- Performs Deduped merge
MERGE INTO store_sales_catalog.silver.family_silver AS tgt
USING (
  SELECT
    family AS family_name,
    current_timestamp() AS created_at,
    current_timestamp() AS updated_at
  FROM (
    SELECT
      family,
      ROW_NUMBER() OVER (
        PARTITION BY family
        ORDER BY current_timestamp()
      ) AS rn
    FROM store_sales_catalog.bronze.train_bronze
    WHERE family IS NOT NULL
  ) sub
  WHERE rn = 1
) AS src
ON tgt.family_name = src.family_name
WHEN MATCHED THEN
  UPDATE SET
    tgt.updated_at = current_timestamp()
WHEN NOT MATCHED THEN
  INSERT (
    family_name,
    created_at,
    updated_at
  )
  VALUES (
    src.family_name,
    src.created_at,
    src.updated_at
  )
```

```python
%sql
MERGE INTO store_sales_catalog.silver.sales_silver AS tgt
USING (
  SELECT
    CAST(date AS DATE) AS date,
    CASE WHEN sales IS NULL OR sales < 0 THEN 0 ELSE sales END AS sales,
    CAST(store_nbr AS INT) AS store_nbr,
    CAST(family AS STRING) AS family,
    COALESCE(onpromotion, 0) AS onpromotion,
    CASE WHEN COALESCE(onpromotion, 0) > 0 THEN TRUE ELSE FALSE END AS is_promotion,
    current_timestamp() AS created_at,
    current_timestamp() AS updated_at,
    id
  FROM (
    SELECT *,
      ROW_NUMBER() OVER (
        PARTITION BY id
        ORDER BY current_timestamp()
      ) AS rn
    FROM store_sales_catalog.bronze.train_bronze
    WHERE date IS NOT NULL
      AND id IS NOT NULL
  ) sub
  WHERE rn = 1
) AS src
ON tgt.sales_id = src.id
WHEN MATCHED THEN
  UPDATE SET
    tgt.date = src.date,
    tgt.sales = src.sales,
    tgt.store_nbr = src.store_nbr,
    tgt.family = src.family,
    tgt.onpromotion = src.onpromotion,
    tgt.is_promotion = src.is_promotion,
    tgt.updated_at = current_timestamp()
WHEN NOT MATCHED THEN
  INSERT (
    date,
    sales,
    store_nbr,
    family,
    onpromotion,
    is_promotion,
    created_at,
    updated_at
  )
  VALUES (
    src.date,
    src.sales,
    src.store_nbr,
    src.family,
    src.onpromotion,
    src.is_promotion,
    src.created_at,
    src.updated_at
  )
```

```python
%sql
-- Incremental merge into silver layer for holidays table
MERGE INTO store_sales_catalog.silver.holidays_events_silver AS tgt
USING (
  SELECT
    date,
    type,
    locale,
    locale_name,
    description,
    CAST(transferred AS BOOLEAN) AS transferred,
    CASE WHEN locale = 'National' THEN TRUE ELSE FALSE END AS is_national,
    CASE WHEN locale = 'Regional' THEN TRUE ELSE FALSE END AS is_regional,
    CASE WHEN locale = 'Local' THEN TRUE ELSE FALSE END AS is_local,
    current_timestamp() AS created_at,
    current_timestamp() AS updated_at
  FROM (
    SELECT *,
      ROW_NUMBER() OVER (
        PARTITION BY date, type, locale, locale_name, description
        ORDER BY current_timestamp()
      ) AS rn
    FROM store_sales_catalog.bronze.holidays_events_bronze
    WHERE date IS NOT NULL
      AND type IS NOT NULL
      AND locale IS NOT NULL
      AND locale_name IS NOT NULL
      AND description IS NOT NULL
      AND transferred IS NOT NULL
  ) sub
  WHERE rn = 1
) AS src
ON tgt.date = src.date
AND tgt.type = src.type
AND tgt.locale = src.locale
AND tgt.locale_name = src.locale_name
AND tgt.description = src.description
WHEN MATCHED THEN
  UPDATE SET
    tgt.transferred = src.transferred,
    tgt.is_national = src.is_national,
    tgt.is_regional = src.is_regional,
    tgt.is_local = src.is_local,
    tgt.locale_name = src.locale_name,
    tgt.description = src.description,
    tgt.updated_at = current_timestamp()
WHEN NOT MATCHED THEN
  INSERT (
    date,
    type,
    locale,
    locale_name,
    description,
    transferred,
    is_national,
    is_regional,
    is_local,
    created_at,
    updated_at
  )
  VALUES (
    src.date,
    src.type,
    src.locale,
    src.locale_name,
    src.description,
    src.transferred,
    src.is_national,
    src.is_regional,
    src.is_local,
    src.created_at,
    src.updated_at
  )
```
